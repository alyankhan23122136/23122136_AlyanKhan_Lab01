```python
booking_status = "Confirmed"
passenger = "Ali"
destination = "Islamabad"

if booking_status == "Confirmed":
    print(f"Ticket confirmed for {passenger} to {destination}. Have a pleasant journey!")
elif booking_status == "Pending":
    print(f"Booking for {passenger} is currently pending. Please wait for confirmation.")
else:
    print(f"Booking for {passenger} has been cancelled. Please contact support to book again.")
```
