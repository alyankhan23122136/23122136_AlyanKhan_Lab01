```python
# Install once in Google Colab
!pip install -U google-genai

from google import genai
from google.colab import userdata

api_key = userdata.get("Gemini_API_Key_FUUAST")

client = genai.Client(api_key=api_key)

MODEL = "models/gemini-3.6-flash"


# =====================================================
# FUUAST_INFO
# University + Semester 6 Section A Library combined
# =====================================================

fuuast_info = {

    "university": {
        "name": "Federal Urdu University of Arts, Science & Technology",
        "short_name": "FUUAST",
        "campus": "Gulshan-e-Iqbal, Karachi",
        "website": "https://www.fuuast.edu.pk",
        "department": {
            "name": "Computer Science",
            "hod": "Dr. Siddiq"
        }
    },

    "my_class": {
        "program": "BSCS",
        "semester": "6th",
        "section": "A",

        "courses": {
            "AI": {
                "name": "Artificial Intelligence",
                "teacher": "Ms. Uzma"
            },
            "NLP": {
                "name": "Natural Language Processing",
                "teacher": "Dr. Khalid"
            },
            "NSC": {
                "name": "Numerical and Symbolic Computing",
                "teacher": "Dr. Akhtar Raza"
            },
            "CC": {
                "name": "Compiler Construction",
                "teacher": "Ms. Naheed"
            },
            "PP": {
                "name": "Professional Practices",
                "teacher": "Not mentioned"
            }
        },

        "timetable": {
            "Monday": {
                "12:20 - 1:10": "NSC - Lab 1",
                "2:00 - 2:50": "NLP - Lab 1",
                "2:50 - 3:40": "AI - Lab 1"
            },
            "Tuesday": {
                "12:20 - 1:10": "CC - Lab 1",
                "2:00 - 2:50": "AI - Lab 1",
                "2:50 - 3:40": "NSC - Lab 1",
                "3:40 - 4:30": "NLP"
            },
            "Wednesday": {
                "12:20 - 1:10": "PP - Room 2",
                "2:00 - 2:50": "NSC - Room 2",
                "2:50 - 3:40": "AI - Lab 1"
            },
            "Thursday": {
                "10:40 - 11:30": "NSC - Lab 1",
                "11:30 - 12:20": "CC - Lab 1",
                "12:20 - 1:10": "PP - Lab 1",
                "2:00 - 2:50": "NLP - Lab 1",
                "2:50 - 3:40": "AI - Lab 1"
            },
            "Friday": {
                "11:30 - 12:20": "CC - Lab 2",
                "12:20 - 1:10": "PP - Room 1"
            }
        }
    }
}


# =====================================================
# ROLE-BASED AGENT
# =====================================================

def fuaast_agent(user_input):

    prompt = f"""
You are FUUAST Assistant, an AI agent for Federal Urdu University,
especially the Computer Science Department.

Use this info as your main source when relevant:

FUUAST_INFO:
{fuuast_info}

Answer questions about Semester 6 Section A, timetable, teachers, courses,
rooms and timings using this info. For anything else - general questions
about the university, or any other topic - just answer normally using
your own knowledge.

Keep answers short and simple.

Question:
{user_input}

Answer:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


# =====================================================
# CHAT LOOP
# =====================================================

print("FUUAST CS Department Agent")
print("BSCS - 6th Semester - Section A")
print("Type 'exit' to stop\n")


while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = fuaast_agent(user)

        print("\nAgent:", reply)
        print()

    except Exception as e:
        print("Error:", e)
```
